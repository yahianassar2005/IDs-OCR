```python
import os
# Fix CPU instruction set conflict
os.environ["PADDLE_PDX_ENABLE_MKLDNN_BYDEFAULT"] = "0"
os.environ["FLAGS_use_mkldnn"] = "0"

import cv2
import sqlite3
from ultralytics import YOLO
# Import the standalone Text Recognizer module from paddleocr / paddlex
from paddleocr import TextRecognition

# --- Configuration (Colab Linux Paths) ---
model_path = r"/content/best v2.pt"
test_image_path = r"/content/flipped_election_gathering-2Fimages_from_30402061500389-2F163419623-2F-2F1ZbRDjTONkn8TtovWsbh-163419623-frontImage_jpg.rf.Av2WZhqoQdRktp6VwrjQ.jpg"
output_dir = r"/content/output/"
db_path = r"/content/national_ids.db"

os.makedirs(output_dir, exist_ok=True)

# --- Initialize Database ---
def init_db(db_path):
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS extracted_ids (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            first_name TEXT,
            full_name TEXT,
            address_1 TEXT,
            address_2 TEXT,
            national_id TEXT,
            source_image TEXT,
            timestamp DATETIME DEFAULT CURRENT_TIMESTAMP
        )
    ''')
    conn.commit()
    conn.close()

init_db(db_path)

# --- Initialize Engines ---
print("Loading Fine-Tuned YOLOv8 and PaddleOCR Engine...")
yolo_model = YOLO(model_path)

# - lang='ar' loads the Arabic recognition model
# - enable_mkldnn=False prevents CPU instruction set crashes
rec_engine = TextRecognition(model_name="arabic_PP-OCRv5_mobile_rec")
print("Models loaded successfully!")

# --- YOLO INFERENCE ---
results = yolo_model.predict(source=test_image_path, conf=0.25, save=False)
result = results[0]

best_predictions = {}
boxes = result.boxes.cpu()

for box in boxes:
    cls_id = int(box.cls[0].item())
    conf = float(box.conf[0].item())
    xyxy = box.xyxy[0].numpy()

    if cls_id not in best_predictions or conf > best_predictions[cls_id][0]:
        best_predictions[cls_id] = (conf, xyxy)

# --- CROP & PADDLEOCR EXTRACTION ---
img_clean = cv2.imread(test_image_path)
h, w, _ = img_clean.shape

extracted_data = {
    "first_name": None,
    "full_name": None,
    "address_1": None,
    "address_2": None,
    "national_id": None
}

class_to_key = {
    0: "first_name",
    1: "full_name",
    2: "address_1",
    3: "address_2",
    4: "national_id"
}

print("\n--- Running PaddleOCR Extraction ---")
for cls_id, (conf, xyxy) in best_predictions.items():
    key = class_to_key.get(cls_id)
    if not key:
        continue

    xmin, ymin, xmax, ymax = map(int, xyxy)

    # Padding settings
    pad_x = 15
    pad_y = 3

    xmin_padded = max(0, xmin - pad_x)
    ymin_padded = max(0, ymin - pad_y)
    xmax_padded = min(w, xmax + pad_x)
    ymax_padded = min(h, ymax + pad_y)

    cropped_field = img_clean[ymin_padded:ymax_padded, xmin_padded:xmax_padded]

    # Smooth upscaling
    crop_h, crop_w = cropped_field.shape[:2]
    target_height = 120
    scale_ratio = target_height / max(1, crop_h)
    target_width = int(crop_w * scale_ratio)

    upscaled = cv2.resize(cropped_field, (target_width, target_height), interpolation=cv2.INTER_LINEAR)

    # Save crop for visual debugging in Colab
    cv2.imwrite(os.path.join(output_dir, f"paddle_crop_{cls_id}_{key}.png"), upscaled)

    # --- RUN PADDLEOCR RECOGNITION ONLY ---
    # Configure recognition-only mode during initialization
   # --- RUN PADDLEOCR RECOGNITION ONLY ---
# Simply pass the image crop without runtime 'det' / 'rec' kwargs
# Pass the upscaled crop directly to the text recognition model
    # Output format: {'res': {'rec_text': 'Text', 'rec_score': 0.98}} or direct string/tuple
    rec_res = rec_engine(upscaled)

    extracted_text = ""
    if rec_res:
        # Extract the prediction safely from the result payload
        if isinstance(rec_res, dict):
            extracted_text = rec_res.get('res', {}).get('rec_text', '')
        elif isinstance(rec_res, list) and len(rec_res) > 0:
            item = rec_res[0]
            if isinstance(item, tuple):
                extracted_text = item[0]
            elif isinstance(item, dict):
                extracted_text = item.get('rec_text', item.get('text', ''))
            else:
                extracted_text = str(item)

    # Post-process National ID digits
    if cls_id == 4:
        extracted_text = str(extracted_text).replace(" ", "").strip()
    else:
        extracted_text = str(extracted_text).strip()

    extracted_data[key] = extracted_text
    print(f"[{key.upper()}]: Detected -> {extracted_text}")
```
